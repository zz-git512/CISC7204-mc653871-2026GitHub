# Activity 9.01 — Plotting Mean Car Prices of Manufacturers

```python
import pandas as pd
from bokeh.io import output_notebook
output_notebook()
```

```python
dataset = pd.read_csv('../Datasets/automobiles.csv')
dataset.head()
```

```python
from bokeh.plotting import figure, show
```

```python
dataset['index'] = dataset.index
```

```python
fig = figure(title='Car prices', x_axis_label='Car Index', y_axis_label='Price')
fig.scatter(dataset['index'], dataset['price'])
show(fig)
```

```python
grouped = dataset.groupby('make', as_index=False).mean(numeric_only=True)
```

```python
grouped_plot = figure(title='Car Manufacturer Mean Prices',
                       x_axis_label='Car Manufacturer',
                       y_axis_label='Mean Price',
                       x_range=grouped['make'])
grouped_plot.scatter(grouped['make'], grouped['price'])
grouped_plot.xaxis.major_label_orientation = 0.785
show(grouped_plot)
```

```python
grouped_plot.xaxis.major_label_orientation = 'vertical'
show(grouped_plot)
```

```python
from bokeh.models import LinearColorMapper
from bokeh.palettes import Magma256

color_mapper = LinearColorMapper(palette=Magma256,
                                 low=grouped['price'].min(),
                                 high=grouped['price'].max())
```

```python
final_plot = figure(title='Car Manufacturer Mean Prices',
                    x_axis_label='Car Manufacturer',
                    y_axis_label='Mean Price',
                    x_range=grouped['make'])
```

```python

from bokeh.models import ColumnDataSource
source = ColumnDataSource(grouped)

final_plot.scatter(x="make",
                   y="price",
                   source=source,
                   size=15,
                   color={"field": "price", "transform": color_mapper})
```

```python
final_plot.xaxis.major_label_orientation = 'vertical'
show(final_plot)
```
