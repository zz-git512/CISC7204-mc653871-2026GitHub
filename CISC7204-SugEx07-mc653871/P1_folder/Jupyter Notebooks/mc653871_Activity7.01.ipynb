# Activity 7.01 – Using Heatmaps to Find Patterns in Flight Passengers' Data




### 02: Import the necessary modules and enable plotting within a Jupyter Notebook.


```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 03: Use pandas to read the `flight_details.csv` dataset located in the Datasets folder. The given dataset contains the monthly figures for flight passengers for the years 1949 to 1960. This dataset originates from the Seaborn library.

The file is stored in the same folder as this notebook, so it can be read directly with the local path `'flight_details.csv'`. The `Months` column is set to a calendar-month order so that the heatmap rows run from January to December.


```python
data = pd.read_csv('flight_details.csv')
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July',
               'August', 'September', 'October', 'November', 'December']
data['Months'] = pd.Categorical(data['Months'], categories=month_order, ordered=True)
```

### 04: Use a heatmap to visualize the given data.


```python
# Reshape the data into a matrix: rows = months, columns = years
data = data.pivot(index='Months', columns='Years', values='Passengers')
```

### 05: Use your own appropriate colormap. Make sure that the lowest value is the brightest, and the highest the darkest, color.

The `GnBu` colormap is a sequential palette that goes from light green (bright) for low values to dark blue for high values, which satisfies the requirement above.


```python
plt.figure(dpi=150)
sns.heatmap(data, cmap='GnBu')
plt.title('Flight Passengers from 1949 to 1960')
plt.show()
```
