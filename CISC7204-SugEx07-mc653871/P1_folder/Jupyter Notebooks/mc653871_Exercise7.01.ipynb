# Exercise 7.01 – Comparing IQ Scores for Different Test Groups by Using a Box Plot



### 02: Import the necessary modules and enable plotting within the `Exercise7.01.ipynb` file.


```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 03: Use the pandas `read_csv()` function to read the data located in the Datasets folder: namely, `"iq_scores.csv"`

The file is stored in the same folder as this notebook, so it can be read directly with the local path `'iq_scores.csv'`.


```python
mydata = pd.read_csv('iq_scores.csv')
```

### 04: Access the data of each test group in the column. Convert this into a list using the `tolist()` method. Once the data of each test group has been converted into a list, assign this list to variables of each respective test group.


```python
group_a = mydata[mydata.columns[0]].tolist()
group_b = mydata[mydata.columns[1]].tolist()
group_c = mydata[mydata.columns[2]].tolist()
group_d = mydata[mydata.columns[3]].tolist()
```

### 05: Print the values of each group to check whether the data inside it is converted into a list. This can be done with the help of the `print()` function.


```python
print(group_a)
print(group_b)
print(group_c)
print(group_d)
```

### 06: Once we have the data for each test group, we need to construct a DataFrame from this data. This can be done with the help of the `pd.DataFrame()` function, which is provided by pandas.


```python
data = pd.DataFrame({'Groups': ['Group A'] * len(group_a) + ['Group B'] * len(group_b) + ['Group C'] * len(group_c) + ['Group D'] * len(group_d),
                     'IQ score': group_a + group_b + group_c + group_d})
data.columns = pd.Index(['Groups', 'IQ score'], dtype=object)  # keep the classic 'object' dtype shown in the sample output
```

### 07: If you don't create your own DataFrame, it is often helpful to print the column names, which is done by calling `print(data.columns)`.


```python
print(data.columns)
```

### 08: Now, since we have the DataFrame, we need to create a box plot using the `boxplot()` function provided by Seaborn. Within this function, specify the variables for both the axes along with the DataFrame. Make Groups the variable to plot on the x-axis, and IQ score the variable for the y-axis. Pass data as a parameter. Moreover, use the whitegrid style, set the context to talk, and remove all axes spines, except the one on the bottom.


```python
plt.figure(dpi=150)
# Set style
sns.set_style('whitegrid')
# Create boxplot (explicit deep palette so each group gets its own colour,
# matching the sample output; hue + legend=False avoids the seaborn 0.13 warning)
sns.boxplot(x='Groups', y='IQ score', data=data, hue='Groups', palette='deep', legend=False)
# Despine
sns.despine(left=True, right=True, top=True)
# Add title
plt.title('IQ scores for different test groups')
# Show plot
plt.show()
```
