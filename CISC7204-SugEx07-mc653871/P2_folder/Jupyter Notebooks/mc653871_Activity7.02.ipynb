# Activity 7.02 – Movie Comparison Revisited



### 01: Use pandas to read the `movie_scores.csv` dataset located in the Datasets folder.




```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

movie_scores = pd.read_csv('movie_scores.csv')
movie_scores.head()
```

### 02: Transform the data into a useable format for Seaborn's `barplot` function.



```python
# Reshape from wide to long format suitable for Seaborn's barplot
scores = movie_scores.melt(id_vars='MovieTitle',
                           value_vars=['AudienceScore', 'Tomatometer'],
                           var_name='Type', value_name='Score')
scores['Type'] = scores['Type'].replace({'AudienceScore': 'Audience Score'})
scores.head()
```

### 03: Use Seaborn to create a visually appealing bar plot that compares the two scores for all five movies.


```python
plt.figure(dpi=150)
sns.barplot(x='MovieTitle', y='Score', hue='Type', data=scores)
plt.title('Movies Scores comparison')
plt.xlabel('Movies')
plt.ylabel('Scores')
plt.xticks(rotation=45)
plt.show()
```
