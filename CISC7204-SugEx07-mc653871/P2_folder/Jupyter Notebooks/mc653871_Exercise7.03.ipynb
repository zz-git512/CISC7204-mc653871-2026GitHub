# Exercise 7.03 – Water Usage Revisited



### 02: Import the necessary modules and enable plotting within the `Exercise7.03.ipynb` file:


```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import squarify
```

### 03: Use the `read_csv()` function of pandas to read the `water_usage.csv` dataset located in the Datasets folder:

The file is stored in the same folder as this notebook, so it is read with the local path `'water_usage.csv'` (with `index_col=0`).


```python
mydata = pd.read_csv('water_usage.csv', index_col=0)
mydata.head()
```

### 04: Create a list of labels by accessing each column from the preceding dataset. Here, the `astype('str')` function is used to cast the fetched data into a type string:


```python
labels = mydata['Usage'] + ' (' + mydata['Percentage'].astype('str') + '%)'
```

### 05: To create a tree map visualization of the given data, use the `plot()` function of the squarify library. This function takes three parameters. The first parameter is a list of all the percentages, and the second parameter is a list of all the labels, which we got in the previous step. The third parameter is the colormap that can be created by using the `light_palette()` function of the Seaborn library:


```python
# Create figure
plt.figure(dpi=200)
# Create tree map
squarify.plot(sizes=mydata['Percentage'],
              label=labels,
              color=sns.light_palette('green', mydata.shape[0]))
plt.axis('off')
# Add title
plt.title('Water usage')
# Show plot
plt.show()
```
