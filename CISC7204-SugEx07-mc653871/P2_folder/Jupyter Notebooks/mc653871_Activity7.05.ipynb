# Activity 7.05 – Linear Regression for Animal Attribute Relations



### 01: Use pandas to read the `anage_data.csv` dataset located in the Datasets folder.



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

mydata = pd.read_csv('anage_data.csv')
mydata.head()
```

### 02: Filter the data so that you end up with samples containing a body mass and maximum longevity. Only consider samples for the `Mammalia` class and a body mass of less than 200,000.


```python
# Keep only Mammalia with a body mass below 200,000
mydata = mydata[(mydata['Class'] == 'Mammalia') & (mydata['Body mass (g)'] < 200000)]

# Keep only the samples that have both a body mass and a maximum longevity
mydata = mydata.dropna(subset=['Body mass (g)', 'Maximum longevity (yrs)'])
mydata.head()
```

### 03: Create a regression plot to visualize the linear relationship between the variables.


```python
plt.figure(dpi=150)
sns.regplot(x='Body mass (g)', y='Maximum longevity (yrs)', data=mydata)
plt.show()
```
