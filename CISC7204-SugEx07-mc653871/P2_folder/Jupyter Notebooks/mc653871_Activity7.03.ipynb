# Activity 7.03 – Comparing IQ Scores for Different Test Groups by Using a Violin Plot




### 01: Use pandas to read the `iq_scores.csv` dataset located in the Datasets folder.




```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

mydata = pd.read_csv('iq_scores.csv')
```

### 02: Access the data of each group in the column, convert it into a list, and assign appropriate variables.


```python
group_a = mydata[mydata.columns[0]].tolist()
group_b = mydata[mydata.columns[1]].tolist()
group_c = mydata[mydata.columns[2]].tolist()
group_d = mydata[mydata.columns[3]].tolist()
```

### 03: Create a pandas DataFrame from the data for each respective group.


```python
data = pd.DataFrame({'Groups': ['Group A'] * len(group_a) + ['Group B'] * len(group_b) + ['Group C'] * len(group_c) + ['Group D'] * len(group_d),
                     'IQ score': group_a + group_b + group_c + group_d})
```

### 04: Create a violin plot for the IQ scores of the different test groups using Seaborn's `violinplot` function.


### 05: Use the `whitegrid` style, set the context to `talk`, and remove all axes spines, except the one on the bottom. Add a title to the plot.


```python
plt.figure(dpi=150)
sns.set_style('whitegrid')
sns.set_context('talk')
sns.violinplot(x='Groups', y='IQ score', data=data,
               hue='Groups', palette='deep', legend=False)
sns.despine(left=True, right=True, top=True)
plt.title('IQ scores for different test groups')
plt.show()
```
