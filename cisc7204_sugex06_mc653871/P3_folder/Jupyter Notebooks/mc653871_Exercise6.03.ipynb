# Exercise 6.03 – Using a Scatter Plot to Visualize Correlation between Various Animals



### 02: Import the necessary modules and enable plotting within the Jupyter Notebook


```python
# Import statements
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

### 03: Use pandas to read the data located in the Datasets folder: namely, "anage_data.csv"


```python
# Load dataset (local copy in the same folder)
data = pd.read_csv('anage_data.csv')
print(data.shape)
```

### 04: Filter the data so that you end up with samples containing a body mass and a maximum longevity

Sort the data according to the animal class; the `isfinite()` function (to check whether the number is finite or not) checks for the finiteness of the given element.


```python
# Preprocessing
longevity = 'Maximum longevity (yrs)'
mass = 'Body mass (g)'
data = data[np.isfinite(data[longevity]) \
    & np.isfinite(data[mass])]
# Sort according to class
amphibia = data[data['Class'] == 'Amphibia']
aves = data[data['Class'] == 'Aves']
mammalia = data[data['Class'] == 'Mammalia']
reptilia = data[data['Class'] == 'Reptilia']
print('Amphibia:', len(amphibia), '| Aves:', len(aves),
      '| Mammalia:', len(mammalia), '| Reptilia:', len(reptilia))
```

### 05: Create a scatter plot visualizing the correlation between the body mass and the maximum longevity

Use different colors to group data samples according to their class. Add a legend, labels, and a title. Use a log scale for both the x-axis and y-axis.


```python
# Create figure
plt.figure(figsize=(10, 6), dpi=300)
# Create scatter plot
plt.scatter(amphibia[mass], amphibia[longevity], \
    label='Amphibia')
plt.scatter(aves[mass], aves[longevity], \
    label='Aves')
plt.scatter(mammalia[mass], mammalia[longevity], \
    label='Mammalia')
plt.scatter(reptilia[mass], reptilia[longevity], \
    label='Reptilia')
# Add legend
plt.legend()
# Log scale
ax = plt.gca()
ax.set_xscale('log')
ax.set_yscale('log')
# Add labels and title
plt.xlabel('Body mass in grams')
plt.ylabel('Maximum longevity in years')
plt.title('Scatter plot on animal statistics')
# Show plot
plt.show()
```
