# Activity 6.07 – Plotting Multiple Images in a Grid




### 01: Import the necessary modules and enable plotting within a Jupyter Notebook


```python
# Import statements
import os
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
%matplotlib inline
```

### 02: Load all four images from the Datasets subfolder: namely, "images.zip"




```python
# Extract the images from the zip archive
import zipfile
with zipfile.ZipFile('images.zip', 'r') as zip_ref:
    zip_ref.extractall()

# Load the four images
img1 = mpimg.imread('images/photo-1.jpg')
img2 = mpimg.imread('images/photo-2.jpg')
img3 = mpimg.imread('images/photo-3.jpg')
img4 = mpimg.imread('images/photo-4.jpg')
print('Loaded', img1.shape, img2.shape, img3.shape, img4.shape)
```

### 03: Visualize the images in a 2x2 grid. Remove the axes and give each image a label.




```python
# Visualize the images in a 2x2 grid
fig = plt.figure(figsize=(8, 8))

plt.subplot(2, 2, 1)
plt.imshow(img1)
plt.title('coast')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(img2)
plt.title('beach')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(img3)
plt.title('building')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(img4)
plt.title('city at night')
plt.axis('off')

plt.suptitle('Visualizing images in a 2x2 grid', y=0.02)
plt.show()
```
