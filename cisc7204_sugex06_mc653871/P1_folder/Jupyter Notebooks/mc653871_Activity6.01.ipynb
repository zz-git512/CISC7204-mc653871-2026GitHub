# Activity 6.01 – Visualizing Stock Trends by Using a Line Plot




### 02: Import the necessary modules and enable plotting within a Jupyter Notebook


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

### 03: Use pandas to read the datasets (GOOGL, FB, AAPL, AMZN, MSFT)


```python
google = pd.read_csv('GOOGL_data.csv')
facebook = pd.read_csv('FB_data.csv')
apple = pd.read_csv('AAPL_data.csv')
amazon = pd.read_csv('AMZN_data.csv')
microsoft = pd.read_csv('MSFT_data.csv')
```

### 04: Use Matplotlib to create a line chart of the closing prices (whole data sequence)

Add labels, a title, and a legend to make the visualization self-explanatory. Use `plt.grid()` to add a grid; rotate the x ticks so the dates are readable.


```python
fig = plt.figure(figsize=(14, 6))

plt.plot(pd.to_datetime(google.date), google.close, label='Google')
plt.plot(pd.to_datetime(facebook.date), facebook.close, label='Facebook')
plt.plot(pd.to_datetime(apple.date), apple.close, label='Apple')
plt.plot(pd.to_datetime(amazon.date), amazon.close, label='Amazon')
plt.plot(pd.to_datetime(microsoft.date), microsoft.close, label='Microsoft')

plt.xlabel('Date')
plt.ylabel('Closing price in $')
plt.title('Stock trend')
plt.legend()
plt.grid()

# Show one date label every 40 trading days (matches the sample output)
n = len(google)
tick_idx = list(range(0, n, 40))
tick_dates = pd.to_datetime(google.date).iloc[tick_idx]
plt.xticks(tick_dates, google.date.iloc[tick_idx], rotation=45)
plt.yticks(np.arange(0, 1401, 100))
plt.ylim(0, 1400)
plt.show()
```
