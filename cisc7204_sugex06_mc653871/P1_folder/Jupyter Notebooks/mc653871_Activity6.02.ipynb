### 02: Import the necessary modules and enable plotting within a Jupyter Notebook


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

### 03: Use pandas to read the dataset 'movie_scores.csv'


```python
movie_scores = pd.read_csv('movie_scores.csv')
```

```python
movie_scores
```

### 04-06: Create a visually appealing bar plot comparing the two scores for all five movies

* Movie titles as x-axis labels; y-axis at intervals of 20 with minor ticks at intervals of 5.
* Add a legend and a suitable title.
* Explicitly specify the axes: `ax = plt.gca()`, `Axes.set_yticks(..., minor=True)`,
  `Axes.yaxis.grid(which='major')` and dashed `Axes.yaxis.grid(which='minor', linestyle='--')`.


```python
movie_scores.plot(kind='bar', x='MovieTitle', y=['Tomatometer', 'AudienceScore'], figsize=(10, 6))

# 06: get a reference to the current axes
ax = plt.gca()

# 05: movie titles as x-axis labels, rotated to stay readable
ax.set_xticklabels(movie_scores.MovieTitle, rotation=45, ha='right')
ax.set_title('Movie comparison')
ax.set_ylabel('Percentage')
ax.set_ylim(0, 100)

# major y-ticks at intervals of 20, minor y-ticks at intervals of 5
ax.set_yticks(range(0, 101, 20))
ax.set_yticks(range(0, 101, 5), minor=True)

# horizontal grid for major ticks and dashed horizontal grid for minor ticks
ax.yaxis.grid(which='major')
ax.yaxis.grid(which='minor', linestyle='--')

# show the y-axis as percentages
from matplotlib.ticker import FuncFormatter
ax.yaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{int(x)}%'))

# legend
ax.legend(['Tomatometer', 'Audience Score'])
plt.show()
```
