### 02: Import the necessary modules and enable plotting within the Jupyter Notebook


```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

### 03: Explicitly create a Figure and set the dpi to 200


### 04: Plot the data pairs (1,1), (2,3), (4,4), (5,3) as circles connected via line segments

Format string `'-o'` = solid line + circle markers.


### 05: Save the plot using the savefig() method



```python
# 03: create the figure with dpi=200
fig = plt.figure(dpi=200)

# 04: plot the data pairs and visualize
plt.plot([1, 2, 4, 5], [1, 3, 4, 3], '-o')
plt.show()

# 05: save the plot
fig.savefig('Exercise6.01.png', bbox_inches='tight')
```
