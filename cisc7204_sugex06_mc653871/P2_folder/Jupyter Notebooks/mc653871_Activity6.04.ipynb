# Activity 6.04 – Comparing Smartphone Sales Units Using a Stacked Area Chart



### 01: Import the necessary modules and enable plotting within a Jupyter Notebook


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

### 02: Use pandas to read the 'smartphone_sales.csv' dataset


```python
# Read the dataset, keeping only the Quarter index and the five manufacturers
sales = pd.read_csv('smartphone_sales.csv', usecols=['Quarter', 'Apple', 'Samsung', 'Huawei', 'Xiaomi', 'OPPO'], index_col='Quarter')
```

```python
sales
```

### 03: Create a visually appealing stacked area chart, with legend, labels, and a title


```python
# Stacked area chart of quarterly sales units (in thousands)
ax = sales.plot(kind='area', stacked=True, figsize=(10, 6))

# Add a legend, labels, and a title
ax.set_title('Smartphone sales units')
ax.set_ylabel('Sales units in thousands')
ax.set_xlabel('Quarters')
ax.legend(title='Manufacturer')
plt.show()
```
