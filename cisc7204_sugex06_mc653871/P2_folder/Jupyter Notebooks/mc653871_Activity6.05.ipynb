### 03: Plot a histogram with 10 bins for the given IQ scores

IQ scores are normally distributed with a mean of 100 and a standard deviation of 15. Visualize the mean as a vertical solid red line, and the standard deviation using dashed vertical lines. Add labels and a title.


```python
# Histogram with 10 bins
plt.figure(figsize=(10, 6))
plt.hist(iq_scores, bins=10, edgecolor='black')

# Mean = 100 (solid red line), mean +/- one std = 85 and 115 (dashed red lines)
plt.axvline(x=100, color='red', linestyle='-', label='Mean')
plt.axvline(x=85, color='red', linestyle='--', label='Mean - 1 SD')
plt.axvline(x=115, color='red', linestyle='--', label='Mean + 1 SD')

plt.xlabel('IQ score')
plt.ylabel('Frequency')
plt.title('IQ scores for a test group of a hundred adults')
plt.legend()
plt.show()
```

### 05: Create a box plot to visualize the same IQ scores. Add labels and a title.


```python
# Box plot of the single test group
plt.figure(figsize=(8, 6))
plt.boxplot(iq_scores)
plt.title('IQ scores for a test group of a hundred adults')
plt.xlabel('Test group')
plt.ylabel('IQ score')
plt.show()
```
