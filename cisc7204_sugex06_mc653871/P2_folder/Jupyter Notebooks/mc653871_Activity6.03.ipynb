# Activity 6.03 – Creating a Stacked Bar Plot to Visualize Restaurant Performance



### 01: Import all the necessary dependencies and load the tips dataset


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# Load the tips dataset (Seaborn built-in)
bills = sns.load_dataset('tips')
```

```python
bills.head()
```

### 02: Create a matrix with the sum of total bills for each day, split by smokers/non-smokers


```python
# Sum of total_bill for each day and smoker status
bill_sums = pd.pivot_table(data=bills, index='day', columns='smoker', values='total_bill', aggfunc=sum)
```

```python
bill_sums
```

### 03-04: Create a stacked bar plot and add legend, labels, and a title


```python
# Stacked bar plot
ax = bill_sums.plot(kind='bar', stacked=True, figsize=(10, 6), rot=0)

# Add legend, labels and a title
ax.set_title('Restaurant performance')
ax.set_xlabel('Day of the week')
ax.set_ylabel('Daily total sales in $')
ax.legend(title='Smoker')
plt.show()
```
