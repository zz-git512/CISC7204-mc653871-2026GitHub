```python
import pandas as pd
```

```python
df = pd.read_csv("forestfires.csv", index_col=0)
```

```python
print(df.head(2))
```

```python
df_filtered = df[df["area"] > 0]
df_filtered.shape
```

```python
area_stats = df_filtered["area"].agg(["mean", "min", "max", "std"])
print(area_stats)
```

```python
df_sorted = df_filtered.sort_values(by="area")
print(df_sorted.tail(20))
```

```python
area_median = df_filtered["area"].median()
print(f"Median of area: {area_median:.2f}")
print(f"Mean of area: {df_filtered['area'].mean():.2f}")
```

```python
unique_months = df["month"].unique()
print(unique_months)
```

```python
march_df = df[df["month"] == "mar"]
print(march_df.shape[0])
```

```python
for month in unique_months:
    month_data = df[df["month"] == month]
    fire_count = month_data.shape[0]
    mean_temp = month_data["temp"].mean()
    print(f"Month: {month}, Number of fires: {fire_count}, Mean temperature: {mean_temp:.2f}")
```
