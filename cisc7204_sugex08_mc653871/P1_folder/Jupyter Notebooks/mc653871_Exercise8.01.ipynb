# Exercise 8.01 - Plotting Poaching Density Using Dot Density and Histograms

In this exercise we use **geoplotlib** to visualize poaching incidents in Tanzania:
1. Dot density map
2. 2D Histogram layer
3. Voronoi diagram

The dataset has 268 rows and 6 columns: `id_report`, `date_report`, `description`, `created_date`, `lat`, `lon`.

## 1. Import dependencies

```python
import geoplotlib
from geoplotlib.utils import read_csv
```

## 2. Load the dataset

```python
dataset = read_csv('poaching_points_cleaned.csv')
```

## 3. Inspect the DataAccessObject

```python
# looking at the dataset structure
dataset
```

## 4. Load with pandas and preview

```python
import pandas as pd
pd_dataset = pd.read_csv('poaching_points_cleaned.csv')
pd_dataset.head()
```

## 5. Dot density map

```python
# plotting our dataset with points
geoplotlib.dot(dataset)
geoplotlib.inline()
```

## 6. Histogram layer

```python
# plotting our dataset as a histogram
geoplotlib.hist(dataset, binsize=20)
geoplotlib.inline()
```

## 7. Voronoi plot

```python
# plotting a voronoi map
geoplotlib.voronoi(dataset, cmap='Blues_r', max_area=1e5, alpha=255)
geoplotlib.inline()
```
