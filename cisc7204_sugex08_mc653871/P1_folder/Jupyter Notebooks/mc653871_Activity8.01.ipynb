# Activity 8.01 - Plotting Geospatial Data on a Map



## 1-2. Import dependencies and load the dataset

```python
import geoplotlib
from geoplotlib.utils import read_csv
import pandas as pd

# load with pandas
df = pd.read_csv('world_cities_pop.csv', low_memory=False)
df.head()
```

## 3. Datatypes and rename columns


```python
# list all datatypes
print(df.dtypes)

# map Latitude -> lat, Longitude -> lon for geoplotlib
df.rename(columns={'Latitude': 'lat', 'Longitude': 'lon'}, inplace=True)
df.head()
```

## 4. Dot density plot of all data points

```python
# plot all cities as dots
geoplotlib.dot(df)
geoplotlib.inline()
```

## 5-6. Aggregate per country


```python
# average population per country
country_group = df.groupby('Country').agg({'Population': 'mean'})
country_group.head(20)
```

```python
# number of cities per country (first 20 entries)
country_count = df.groupby('Country').size().sort_values(ascending=False)
country_count.head(20)
```

## 7. Dot plot of the reduced dataset

```python
# keep only rows with a recorded population (> 0)
df_reduced = df[df['Population'] > 0]
print('rows after pop>0 filter:', len(df_reduced))
geoplotlib.dot(df_reduced)
geoplotlib.inline()
```

## 8. Filter cities with population > 100,000

```python
df_pop = df[df['Population'] > 100000]
print('cities with pop > 100000:', len(df_pop))
df_pop.head()
```

## 9. Voronoi tessellation layer

```python
# Voronoi density of large cities, fixed to a world view
from geoplotlib.utils import BoundingBox
geoplotlib.set_bbox(BoundingBox.WORLD)
geoplotlib.voronoi(df_pop, cmap='hot_r', max_area=1e5, alpha=255)
geoplotlib.inline()
```

## 10. Filter to Germany (de) and Great Britain (gb)

```python
df_gb = df_pop[df_pop['Country'].isin(['de', 'gb'])]
print('Germany + GB large cities:', len(df_gb))
df_gb.head()
```

## 11. Delaunay triangulation layer

```python
# Delaunay triangulation reveals the most densely populated corridors
geoplotlib.delaunay(df_gb, cmap='hot_r', line_width=2)
geoplotlib.inline()
```
