## 1. Import dependencies

```python
import geoplotlib
```

## 2. Default tile provider (CartoDB Positron)

```python
# display the map with the default tile provider (CartoDB Positron)
geoplotlib.inline()
```

## 3. Dark Matter tile provider

```python
# using map tiles from the dark matter tile provider
geoplotlib.tiles_provider('darkmatter')
geoplotlib.inline()
```

## 4. Custom tile provider (OSM France / humanitarian)

```python
geoplotlib.tiles_provider({
    'url': lambda zoom, xtile, ytile:
        'https://a.tile.openstreetmap.fr/'
        'hot/%d/%d/%d.png' % (zoom, xtile, ytile),
    'tiles_dir': 'custom_tiles',
    'attribution': 'Custom Tiles Provider - Humanitarian map style'
})
geoplotlib.inline()
```
