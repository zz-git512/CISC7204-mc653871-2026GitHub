# Exercise 8.02 - Creating a Choropleth Plot with GeoJSON Data



## 1. Import dependencies

```python
# importing the necessary dependencies
import json
import geoplotlib
from geoplotlib.colors import ColorMap
from geoplotlib.utils import BoundingBox
```

## 2. Inspect the 4th entry of the states dataset

```python
# displaying the fourth entry of the states dataset
with open('us_states.json') as data:
    dataset = json.load(data)

fourth_state = dataset.get('features')[3]

# only showing one coordinate instead of all points
fourth_state['geometry']['coordinates'] = \
    fourth_state['geometry']['coordinates'][0][0]
print(json.dumps(fourth_state, indent=4))
```

## 3. List all US states

```python
# listing the states in the dataset
with open('us_states.json') as data:
    dataset = json.load(data)

states = [feature['properties']['NAME'] for feature in dataset.get('features')]

print('Number of states:', len(states))
print(states)
```

## 4. Plot the state outlines

```python
# plotting the information from the geojson file
geoplotlib.geojson('us_states.json')
geoplotlib.set_bbox(BoundingBox.USA)
geoplotlib.show()
```

## 5. Build the ColorMap for CENSUSAREA

```python
cmap = ColorMap('Reds', alpha=255, levels=10)

def get_color(properties):
    return cmap.to_color(properties['CENSUSAREA'], maxvalue=300000, scale='lin')
```

## 6. Choropleth: shaded states + black outlines, focus on USA

```python
"""
plotting the shaded states and adding another layer which plots the state outlines in black
our BoundingBox should focus the USA
"""
geoplotlib.geojson('us_states.json', fill=True, color=get_color)
geoplotlib.geojson('us_states.json', fill=False, color=[0, 0, 255])

geoplotlib.set_bbox(BoundingBox.USA)
geoplotlib.show()
```

## 7. Add tooltip with state name and census area

```python
# adding the f_tooltip that displays name and census area on hover
geoplotlib.geojson('us_states.json',
                fill=True, color=get_color,
                f_tooltip=lambda properties: properties['NAME']
                + ' - Census Areas: ' \
                + str(properties['CENSUSAREA']))
geoplotlib.geojson('us_states.json', fill=False, color=[0, 0, 255])

geoplotlib.set_bbox(BoundingBox.USA)
geoplotlib.show()
```
